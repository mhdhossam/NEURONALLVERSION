```python
def load_dataset(folder_path):
    X, y = [], []
    for person_name in os.listdir(folder_path):
        person_path = os.path.join(folder_path, person_name)
        if os.path.isdir(person_path):
            for img_file in os.listdir(person_path):
                if img_file.endswith(('.jpg', '.png')):
                    emb = get_face_embedding(os.path.join(person_path, img_file))
                    if emb is not None:
                        X.append(emb)
                        y.append(person_name)
    return np.array(X), np.array(y)
```

```python
#------------------------download logic---------------------
def safe_load_image(url):
    try:
        headers = {"User-Agent": "Mozilla/5.0"}
        r = requests.get(url, headers=headers, timeout=10)

        if r.status_code != 200:
            return None

        if "image" not in r.headers.get("Content-Type", ""):
            return None

        img = Image.open(BytesIO(r.content)).convert("RGB")
        return img

    except:
        return None
```

```python
class CandidateIdentity:
    def __init__(self, name):
        self.id = np.random.randint(1e6, 1e9)
        self.name = name
        self.confidence = 0.0
        self.sources = []
        self.embeddings = []
        self.status = "PENDING"

    def add_source(self, url):
        if url not in self.sources:
            self.sources.append(url)

    def add_embedding(self, embedding):
        self.embeddings.append(embedding)

    def update_confidence(self, delta):
        self.confidence = max(0.0, min(1.0, self.confidence + delta))

    def promote(self):
        self.status = "VERIFIED"
```

```python
# ---------------------------- Neuron for Face ----------------------------
class Neuron:
    def __init__(self, embedding, label):
        self.id = np.random.randint(1e6, 1e9)
        self.embedding = embedding
        self.label = label

        self.confidence = 0.5
        self.support = 1
        self.mistakes = 0
        self.links = {}

    def update_confidence(self, correct: bool):
        if correct:
            self.support += 1
            self.confidence = min(
                1.0,
                self.confidence + (1.0 / self.support)
            )
        else:
            self.mistakes += 1
            self.confidence = max(
                0.0,
                self.confidence - 0.1
            )

    def increment_mistakes(self):
        self.mistakes += 1
    @staticmethod
    def cosine_similarity(a, b):
        return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b) + 1e-10)

    def decay(self, rate=0.99):
        self.confidence *= rate
```

```python
# ---------------------------- Neuron Network ----------------------------
class NeuronNetwork:
    def __init__(self, db_path=None):
        self.neurons = {}
        self.candidates = {}
        self.db_path = db_path
        if db_path:
            self._init_db()
    def compute_reward(self, correct, similarity):
        """
        RL reward function:
        - correctness matters most
        - similarity adjusts strength
        """

        if correct is None:
            return 0.0

        base = 1.0 if correct else -1.0

        # similarity amplifies learning signal
        return base * (0.5 + similarity)
    def create_candidate(self, name="UNKNOWN"):
        candidate = CandidateIdentity(name)
        self.candidates[candidate.id] = candidate
        return candidate.id


    def get_candidate(self, cid):
        return self.candidates.get(cid)


    def remove_candidate(self, cid):
        if cid in self.candidates:
            del self.candidates[cid]


    def promote_candidate(self, cid):
        candidate = self.candidates.get(cid)

        if not candidate:
            return None

        if len(candidate.embeddings) == 0:
            return None

        avg_embedding = np.mean(
            candidate.embeddings,
            axis=0
        )

        nid = self.create_neuron(
            avg_embedding,
            candidate.name
        )

        del self.candidates[cid]

        return nid
    def print_candidates(self):

        print("\n===== CANDIDATES =====")

        for cid, candidate in self.candidates.items():

            print(
                f"ID={cid} | "
                f"Name={candidate.name} | "
                f"Confidence={candidate.confidence:.2f} | "
                f"Embeddings={len(candidate.embeddings)} | "
                f"Sources={len(candidate.sources)} | "
                f"Status={candidate.status}"
            )
        
    @staticmethod
    def cosine_similarity(a, b):
        return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b) + 1e-10)

    def create_neuron(self, embedding, label):
        n = Neuron(embedding, label)
        self.neurons[n.id] = n
        # Link similar neurons
        for nid, neuron in self.neurons.items():
            if nid == n.id: 
                continue
            sim = self.cosine_similarity(n.embedding, neuron.embedding)
            if sim > 0.7:
                self.link_neurons(n.id, nid, weight=sim)
        return n.id

    def find_similar(self, embedding, top_k=3, threshold=0.6):
        scored = []

        for nid, neuron in self.neurons.items():
            sim = self.cosine_similarity(neuron.embedding, embedding)

            if sim >= threshold:
                scored.append((nid, sim))

        # sort best matches first
        scored.sort(key=lambda x: x[1], reverse=True)

        # return top-k results
        return scored[:top_k] if scored else None

    def update_neuron(self, nid, correct=True, new_embedding=None, similarity=0.0):

        neuron = self.neurons[nid]

        reward = self.compute_reward(correct, similarity)

        # ----------------------------
        # RL-style confidence update
        # ----------------------------
        neuron.confidence += reward * 0.2
        neuron.confidence = max(0.0, min(1.0, neuron.confidence))

        # ----------------------------
        # mistake tracking
        # ----------------------------
        if reward < 0:
            neuron.increment_mistakes()

        # ----------------------------
        # embedding smoothing (learning)
        # ----------------------------
        if new_embedding is not None:
            alpha = 0.1 + (0.2 * abs(reward))  # stronger reward = stronger update
            neuron.embedding = (1 - alpha) * neuron.embedding + alpha * new_embedding
    def link_neurons(self, nid1, nid2, weight=1.0):
        self.neurons[nid1].links[nid2] = weight
        self.neurons[nid2].links[nid1] = weight

    # Decay and prune
    def decay_and_prune(self, min_confidence=0.1):
        to_delete = []
        for nid, neuron in self.neurons.items():
            neuron.decay()
            if neuron.confidence < min_confidence:
                to_delete.append(nid)
        for nid in to_delete:
            for linked_id in list(self.neurons[nid].links.keys()):
                if linked_id in self.neurons:
                    self.neurons[linked_id].links.pop(nid, None)
            self.neurons.pop(nid)

    # ---------------------------- SQLite persistence ----------------------------
    def _init_db(self):
        self.conn = sqlite3.connect(self.db_path, timeout=10)
        self.conn.execute("PRAGMA journal_mode=WAL;")  # better concurrency
        c = self.conn.cursor()
        c.execute('''
            CREATE TABLE IF NOT EXISTS neurons (
                id INTEGER PRIMARY KEY,
                label TEXT,
                confidence REAL,
                mistakes INTEGER,
                embedding TEXT,
                links TEXT
            )
        ''')
        self.conn.commit()

    import time

    def save(self, retries=5):
        if not self.db_path:
            raise ValueError("No db_path set for saving.")

        for attempt in range(retries):
            try:
                c = self.conn.cursor()
                for nid, neuron in self.neurons.items():
                    emb_json = json.dumps(neuron.embedding.tolist())
                    links_dict = {str(k): float(v) for k, v in neuron.links.items()}
                    links_json = json.dumps(links_dict)

                    c.execute('''
                        INSERT OR REPLACE INTO neurons (id, label, confidence, mistakes, embedding, links)
                        VALUES (?, ?, ?, ?, ?, ?)
                    ''', (nid, neuron.label, neuron.confidence, neuron.mistakes, emb_json, links_json))

                self.conn.commit()
                return  # success

            except sqlite3.OperationalError as e:
                if "locked" in str(e):
                    print(f"⚠️ DB locked, retrying... ({attempt+1})")
                    time.sleep(1)
                else:
                    raise e
    def close(self):
        if hasattr(self, "conn"):
            self.conn.close()

    def load(self):
        if not self.db_path:
            raise ValueError("No db_path set for loading.")
        c = self.conn.cursor()
        c.execute('SELECT id, label, confidence, mistakes, embedding, links FROM neurons')
        rows = c.fetchall()
        self.neurons = {}
        for row in rows:
            nid, label, conf, mistakes, emb_json, links_json = row
            neuron = Neuron(np.array(json.loads(emb_json)), label)
            neuron.id = nid
            neuron.confidence = conf
            neuron.mistakes = mistakes
            neuron.links = json.loads(links_json)
            self.neurons[nid] = neuron

    def print_summary(self):
        for nid, n in self.neurons.items():
            print(f"Neuron {nid}: label={n.label}, conf={n.confidence:.2f}, mistakes={n.mistakes}, links={list(n.links.keys())}")
```

```python
def enrich_candidate_real(nn, cid, search_results, mtcnn, resnet, device):

    candidate = nn.get_candidate(cid)

    if not candidate:
        return

    for r in search_results:

        img = safe_load_image(r)

        if img is None:
            continue

        emb = image_to_embedding(img, mtcnn, resnet, device)

        if emb is None:
            continue

        # store real evidence
        candidate.add_source(r)
        candidate.add_embedding(emb)
        candidate.update_confidence(0.2)

    print(f"\n===== REAL ENRICHMENT DONE =====")
    print(f"Sources: {len(candidate.sources)}")
    print(f"Embeddings: {len(candidate.embeddings)}")
    print(f"Confidence: {candidate.confidence:.2f}")
    
def try_promote_candidate(nn, cid, threshold=0.75):

    candidate = nn.get_candidate(cid)

    if not candidate:
        return None

    if candidate.confidence >= threshold and len(candidate.embeddings) >= 3:

        avg_embedding = np.mean(candidate.embeddings, axis=0)

        nid = nn.create_neuron(avg_embedding, candidate.name)

        candidate.promote()

        nn.remove_candidate(cid)

        print(f"🔥 Candidate promoted to neuron: {nid}")

        return nid

    return None
    
```

```python
def train_svm(X_train, y_train):
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X_train)
    clf = SVC(kernel='rbf', probability=True)
    clf.fit(X_scaled, y_train)
    return clf, scaler
```

```python
def decide_action(result, neuron_candidates=None):
    
    svm_conf = result["svm_confidence"]
    similarity = result["similarity"]

    # ----------------------------
    # BASE SCORE (SVM strength)
    # ----------------------------
    svm_score = svm_conf * 0.5

    # ----------------------------
    # NEURON SCORE (memory strength)
    # ----------------------------
    neuron_score = 0.0

    if neuron_candidates:
        sims = [sim for _, sim in neuron_candidates]
        if sims:
            neuron_score = max(sims) * 0.5

    # ----------------------------
    # FINAL SCORE
    # ----------------------------
    total_score = svm_score + neuron_score

    # ----------------------------
    # DECISION POLICY
    # ----------------------------
    if total_score >= 0.85:
        return "TRUST"

    elif total_score >= 0.55:
        return "VERIFY"

    else:
        return "SEARCH"
```

```python
def image_to_embedding(img, mtcnn, resnet, device):
    face = mtcnn(img)

    if face is None:
        return None

    face = face.unsqueeze(0).to(device)

    with torch.no_grad():
        emb = resnet(face)

    return emb.squeeze(0).cpu().numpy()
```

```python
def predict_face(nn: NeuronNetwork, svm_clf, scaler, img_path, true_label=None):
    emb = get_face_embedding(img_path)

    if emb is None:
        return {
            "prediction": "NO_FACE",
            "svm_confidence": 0.0,
            "neuron_id": None,
            "similarity": 0.0,
        }

    # ----------------------------
    # SVM Prediction
    # ----------------------------
    scaled_emb = scaler.transform([emb])
    pred_label = svm_clf.predict(scaled_emb)[0]
    probs = svm_clf.predict_proba(scaled_emb)[0]

    max_prob = float(np.max(probs))

    if max_prob < 0.60:
        pred_label = "UNKNOWN"

    # ----------------------------
    # TOP-K NEURON SEARCH
    # ----------------------------
    candidates = nn.find_similar(emb)

    best_nid = None
    best_score = 0.0
    best_similarity = 0.0

    if candidates:
        for nid, sim in candidates:
            neuron = nn.neurons[nid]
            score = sim * neuron.confidence

            if score > best_score:
                best_score = score
                best_nid = nid
                best_similarity = sim

    matched_nid = best_nid

    if not candidates:
        matched_nid = None
        best_similarity = 0.0

    # ----------------------------
    # RL / Brain Update
    # ----------------------------

    correct = None
    if true_label is not None:
        correct = (pred_label == true_label)

    # UNKNOWN CASE
    if pred_label == "UNKNOWN":

        if matched_nid is not None:
            nn.update_neuron(
                matched_nid,
                correct=False,
                new_embedding=emb
            )
        else:
            matched_nid = nn.create_neuron(
                emb,
                "UNKNOWN"
            )

        return {
            "prediction": "UNKNOWN",
            "svm_confidence": max_prob,
            "neuron_id": matched_nid,
            "similarity": best_similarity,
        }

    # KNOWN CASE
    if matched_nid is not None and correct is not None:
        nn.update_neuron(
            matched_nid,
            correct=correct,
            new_embedding=emb
        )
    

    return {
        "prediction": pred_label,
        "svm_confidence": max_prob,
        "neuron_id": matched_nid,
        "similarity": best_similarity,
    }
```

```python
def upload_to_imgbb(image_path, api_key):
    import base64
    import requests

    with open(image_path, "rb") as f:
        b64 = base64.b64encode(f.read()).decode()

    r = requests.post(
        "https://api.imgbb.com/1/upload",
        data={"key": api_key, "image": b64}
    )

    data = r.json()

    # DEBUG (VERY IMPORTANT)
    if "data" not in data:
        print("IMGBB ERROR RESPONSE:", data)
        raise Exception("ImgBB upload failed")

    return data["data"]["url"]
```

```python

async def yandex_search_image(example_img, output_path):

    async with async_playwright() as p:
        browser = await p.chromium.launch(headless=True)
        page = await browser.new_page()

        try:
            await page.goto("https://yandex.com/images/")
            await page.wait_for_timeout(2000)

            file_input = page.locator('input[type="file"]').first
            await file_input.set_input_files(example_img)

            await page.wait_for_timeout(5000)

            text = await page.inner_text("body")

            with open(output_path, "w", encoding="utf-8") as f:
                f.write(text)

            return text

        finally:
            await browser.close()


# ----------------------------
# NAME EXTRACTION (NO LLM)
# ----------------------------
def extract_name(text):

    names = re.findall(
        r"\b[A-Z][a-z]+(?:\s+[A-Z][a-z]+)+\b",
        text or ""
    )

    if not names:
        return None

    return Counter(names).most_common(1)[0][0]


# ----------------------------
# MAIN SEARCH PIPELINE
# ----------------------------
def run_search_pipeline(nn, example_img, IMGBB_KEY, upload_to_imgbb, enrich_candidate_real, try_promote_candidate, mtcnn, resnet, device):

    output_path =  r"C:\Users\mhd_gamer\Documents\GitHub\face_recognition\yandex_output.txt"

    # upload image
    img_url = upload_to_imgbb(example_img, IMGBB_KEY)

    # run yandex (ONLY ONCE)
    loop = asyncio.ProactorEventLoop()
    asyncio.set_event_loop(loop)
    nest_asyncio.apply(loop)
    text = loop.run_until_complete(
        yandex_search_image(example_img, output_path)
    )

    # extract name
    name_hint = extract_name(text)

    print("NAME HINT:", name_hint)

    # create candidate
    cid = nn.create_candidate(name=name_hint or "UNKNOWN")

    # enrich candidate
    search_results = []  # we only use text-based pipeline now

    enrich_candidate_real(
        nn,
        cid,
        search_results,
        mtcnn,
        resnet,
        device
    )

    # promote if possible
    promoted_id = try_promote_candidate(nn, cid)

    if promoted_id:
        print(f"🔥 Promoted to neuron: {promoted_id}")
    else:
        print("⏳ Still learning candidate...")


# ----------------------------
# USAGE EXAMPLE
# ----------------------------
if __name__ == "__main__":
   

    test_folder = r"C:\Users\mhd_gamer\Documents\GitHub\face_recognition\FACE DATASET\Faces\test"
    example_img = os.path.join(test_folder, "Elon_Musk_-_54820081119_(cropped).jpg")

    result = predict_face(nn, svm_clf, scaler, example_img)

    action = decide_action(result)

    search_results = []

    if action == "SEARCH":

        print("\n===== SEARCH TRIGGERED =====")

        IMGBB_KEY = "7ddcd1176db91fb7b66ff5b0533eabb8"

        run_search_pipeline(
            nn,
            example_img,
            IMGBB_KEY,
            upload_to_imgbb,
            enrich_candidate_real,
            try_promote_candidate,
            mtcnn,
            resnet,
            device
        )

    # ----------------------------
    # OUTPUT
    # ----------------------------
    print("\n===== BRAIN DECISION =====")
    print(f"Prediction     : {result['prediction']}")
    print(f"SVM Confidence : {result['svm_confidence']:.4f}")
    print(f"Similarity     : {result['similarity']:.4f}")
    print(f"Neuron ID      : {result['neuron_id']}")
    print(f"Action         : {action}")

```
