```python
from dataclasses import dataclass, field
from uuid import uuid4
from typing import List, Tuple, Dict, Set, Any
from collections import defaultdict
import heapq
import os
import numpy as np
import torch
from PIL import Image
from facenet_pytorch import MTCNN, InceptionResnetV1


# =========================
# SOMA (Cell Body)
# =========================

@dataclass
class Soma:
    uuid: str = field(default_factory=lambda: str(uuid4()))
    data: Dict[str, Any] = field(default_factory=dict)
    metadata: Dict[str, Any] = field(default_factory=dict)
    confidence: float = 0.5
    support: int = 1
    mistakes: int = 0


# =========================
# DENDRITES (Inputs)
# =========================

@dataclass
class Dendrite:
    uuid: str = field(default_factory=lambda: str(uuid4()))
    input_neurons: List[str] = field(default_factory=list)
    signal_buffer: List[Any] = field(default_factory=list)

    def receive(self, signal):
        self.signal_buffer.append(signal)


# =========================
# AXON (Outputs)
# =========================

@dataclass
class Axon:
    uuid: str = field(default_factory=lambda: str(uuid4()))
    output_neurons: List[str] = field(default_factory=list)
    output_buffer: List[Any] = field(default_factory=list)

    def send(self, signal):
        self.output_buffer.append(signal)


# =========================
# SYNAPSE
# =========================

@dataclass
class Synapse:
    uuid: str = field(default_factory=lambda: str(uuid4()))
    source: str = ""
    target: str = ""
    weight: float = 1.0
    confidence: float = 0.5
    usage: int = 0
    decay_rate: float = 0.99

    def reinforce(self, reward: float, learning_rate: float = 0.1):
        delta = reward * learning_rate
        self.weight += delta
        self.confidence += delta * 0.5
        self.usage += 1
        self.weight = max(0.0, min(self.weight, 10.0))
        self.confidence = max(0.0, min(self.confidence, 1.0))

    def decay(self):
        self.weight *= self.decay_rate
        self.confidence *= self.decay_rate
        self.weight = max(0.01, self.weight)
        self.confidence = max(0.01, self.confidence)


# =========================
# NEURON
# =========================

class Neuron:

    def __init__(self):
        self.uuid = str(uuid4())
        self.soma = Soma()
        self.dendrites = Dendrite()
        self.axon = Axon()
        self.synapses: Dict[str, Synapse] = {}
        self.activation = 0.0
        self.last_active_step = 0

    def activate_neuron(neuron, strength: float):
        neuron.activation += strength
        neuron.activation = min(1.0, neuron.activation)
        neuron.last_active_step += 1

    def add_synapse(self, target_neuron_id, weight=1.0):
        synapse = Synapse(
            source=self.uuid,
            target=target_neuron_id,
            weight=weight
        )
        self.synapses[synapse.uuid] = synapse
        self.axon.output_neurons.append(target_neuron_id)
        return synapse.uuid


# =========================
# NEURAL PLATFORM
# =========================

@dataclass
class NeuralPlatform:
    uuid: str = field(default_factory=lambda: str(uuid4()))
    name: str = ""
    metadata: Dict[str, Any] = field(default_factory=dict)
    region_refs: Set[str] = field(default_factory=set)
    platform_links: Set[str] = field(default_factory=set)

    def add_region(self, region_uuid: str):
        self.region_refs.add(region_uuid)

    def remove_region(self, region_uuid: str):
        self.region_refs.discard(region_uuid)

    def link_platform(self, other_platform_uuid: str):
        if other_platform_uuid != self.uuid:
            self.platform_links.add(other_platform_uuid)

    def unlink_platform(self, other_platform_uuid: str):
        self.platform_links.discard(other_platform_uuid)


# =========================
# REGION
# =========================

@dataclass
class Region:
    uuid: str = field(default_factory=lambda: str(uuid4()))
    name: str = ""
    metadata: Dict[str, Any] = field(default_factory=dict)
    platforms: Set[str] = field(default_factory=set)
    neurons: Set[str] = field(default_factory=set)
    signature: Dict[str, float] = field(default_factory=dict)
    activation: float = 0.0

    def add_platform(self, platform_uuid: str):
        self.platforms.add(platform_uuid)

    def remove_platform(self, platform_uuid: str):
        self.platforms.discard(platform_uuid)

    def has_platform(self, platform_uuid: str) -> bool:
        return platform_uuid in self.platforms

    def add_neuron(self, neuron_uuid: str):
        self.neurons.add(neuron_uuid)

    def remove_neuron(self, neuron_uuid: str):
        self.neurons.discard(neuron_uuid)

    def has_neuron(self, neuron_uuid: str) -> bool:
        return neuron_uuid in self.neurons

    def activate(self, strength: float = 1.0):
        self.activation += strength
        self.activation = min(1.0, self.activation)

    def decay(self, rate: float = 0.98):
        self.activation *= rate

    def similarity_score(self, query_signature: Dict[str, float]) -> float:
        score = 0.0
        for k, v in query_signature.items():
            score += v * self.signature.get(k, 0.0)
        return score


# =========================
# NEURON SEARCH ENGINE
# =========================

class NeuronSearchEngine:

    def __init__(self, platforms, regions, neurons):
        self.platforms = platforms
        self.regions = regions
        self.neurons = neurons

    def resolve_entry(self, platform_uuid: str) -> Set[str]:
        platform = self.platforms.get(platform_uuid)
        if not platform:
            return set()
        return platform.region_refs

    def expand_regions(self, region_ids: Set[str]) -> Set[str]:
        neurons = set()
        for rid in region_ids:
            region = self.regions.get(rid)
            if not region:
                continue
            region.activate(0.5)
            neurons.update(region.neurons)
        return neurons

    def traverse(self, neuron_ids: Set[str], depth=2):
        visited = set(neuron_ids)
        frontier = set(neuron_ids)
        for _ in range(depth):
            next_frontier = set()
            for nid in frontier:
                neuron = self.neurons.get(nid)
                if not neuron:
                    continue
                for syn in neuron.synapses.values():
                    target = syn.target
                    if target not in visited:
                        visited.add(target)
                        next_frontier.add(target)
            frontier = next_frontier
        return visited

    def score(self, neuron):
        soma = neuron.soma
        return (
            soma.confidence +
            (soma.support * 0.01) -
            (soma.mistakes * 0.05) +
            (len(neuron.synapses) * 0.01)
        )

    def search(self, platform_uuid: str, top_k=5):
        region_ids = self.resolve_entry(platform_uuid)
        neuron_pool = self.expand_regions(region_ids)
        expanded = self.traverse(neuron_pool, depth=2)
        scored = []
        for nid in expanded:
            neuron = self.neurons.get(nid)
            if not neuron:
                continue
            score = self.score(neuron)
            heapq.heappush(scored, (-score, nid))
        results = []
        for _ in range(min(top_k, len(scored))):
            results.append(heapq.heappop(scored)[1])
        return results


# =========================
# BRAIN SYSTEM
# =========================

class BrainSystem:

    def __init__(self):
        self.neurons = {}
        self.regions = {}
        self.platforms = {}
        self.synapses = {}

    def create_neuron(self):
        nid = str(uuid4())
        self.neurons[nid] = Neuron()
        return nid

    def add_region(self, region: Region):
        self.regions[region.uuid] = region


# =========================
# HELPER FUNCTIONS
# =========================

def propagate(neurons: dict, start_ids: set, steps: int = 3):
    active = {nid: 1.0 for nid in start_ids}
    for _ in range(steps):
        next_active = {}
        for nid, activation in active.items():
            neuron = neurons.get(nid)
            if not neuron:
                continue
            neuron.activation = min(1.0, neuron.activation + activation)
            for syn in neuron.synapses.values():
                target = neurons.get(syn.target)
                if not target:
                    continue
                signal = activation * syn.weight * syn.confidence
                if signal < 0.01:
                    continue
                if syn.target not in next_active:
                    next_active[syn.target] = 0.0
                next_active[syn.target] += signal
        active = next_active


def decay_network(neurons: dict, rate: float = 0.95):
    for neuron in neurons.values():
        neuron.activation *= rate
        neuron.soma.confidence *= rate


def activation_search(neurons: dict, top_k=5):
    scored = []
    for nid, neuron in neurons.items():
        if neuron.activation <= 0:
            continue
        score = neuron.activation * 0.6 + neuron.soma.confidence * 0.4
        scored.append((score, nid))
    scored.sort(reverse=True)
    return [nid for _, nid in scored[:top_k]]


def get_best_paths(neuron, neurons, top_k=3):
    candidates = []
    for syn in neuron.synapses.values():
        target = neurons.get(syn.target)
        if not target:
            continue
        score = syn.weight * 0.6 + syn.confidence * 0.4
        candidates.append((score, target.uuid))
    candidates.sort(reverse=True)
    return candidates[:top_k]


def update_synapse(synapse, correct: bool):
    if correct:
        synapse.reinforce(+1.0)
    else:
        synapse.reinforce(-1.0)
    synapse.decay()


def get_candidate_neurons(nn, platform):
    candidate_regions = platform.region_refs
    candidates = set()
    for rid in candidate_regions:
        region = nn.regions.get(rid)
        if not region:
            continue
        region.activate(0.3)
        candidates.update(region.neurons)
    return candidates


# -------------------------
# Metadata-selection guard
# -------------------------
# Image embeddings live in this separate feature store. The graph stores identity
# metadata only; embeddings are never copied into Soma.data or Soma.metadata.
FEATURES_BY_NEURON = {}
NEURON_METADATA_MIN_SIMILARITY = 0.60
NEURON_METADATA_MIN_MARGIN = 0.025
NEURON_METADATA_TOP_EXEMPLARS = 3
LAST_METADATA_DECISION = {}


def _normalise_embedding(vector):
    arr = np.asarray(vector, dtype=np.float32).reshape(-1)
    if arr.size == 0 or not np.isfinite(arr).all():
        return None
    norm = float(np.linalg.norm(arr))
    if not np.isfinite(norm) or norm <= 1e-12:
        return None
    return arr / norm


def find_best_in_candidates(nn, embedding, candidates):
    """Select metadata only when visual evidence is strong and unambiguous.

    Similarity is ranked independently of metadata confidence/activation. Each
    identity is represented by the mean of its top few training-example
    similarities, which is more robust than one repeatedly smoothed prototype.
    """
    global LAST_METADATA_DECISION
    query = _normalise_embedding(embedding)
    if query is None:
        LAST_METADATA_DECISION = {"status": "rejected", "reason": "invalid_query_embedding"}
        return None, 0.0

    ranked = []
    for nid in sorted(set(candidates), key=str):
        neuron = nn.neurons.get(nid)
        if neuron is None:
            continue
        label = neuron.soma.metadata.get("label")
        if not isinstance(label, str) or not label.strip():
            # Missing metadata must never be turned into a guessed label.
            continue

        exemplars = FEATURES_BY_NEURON.get(nid, ())
        similarities = []
        for exemplar in exemplars:
            stored = np.asarray(exemplar, dtype=np.float32).reshape(-1)
            if stored.shape != query.shape or not np.isfinite(stored).all():
                continue
            similarity = float(np.dot(query, stored))
            if np.isfinite(similarity):
                similarities.append(similarity)

        if not similarities:
            continue

        top_similarities = sorted(similarities, reverse=True)[:NEURON_METADATA_TOP_EXEMPLARS]
        class_score = float(sum(top_similarities) / len(top_similarities))
        ranked.append({
            "neuron_id": nid,
            "label": label,
            "score": class_score,
            "best_example_similarity": max(similarities),
            "examples": len(similarities),
        })

    ranked.sort(key=lambda row: (-row["score"], row["label"].casefold(), str(row["neuron_id"])))
    if not ranked:
        LAST_METADATA_DECISION = {"status": "rejected", "reason": "no_valid_metadata_candidates"}
        return None, 0.0

    best = ranked[0]
    runner_up = ranked[1] if len(ranked) > 1 else None
    runner_score = runner_up["score"] if runner_up else None
    margin = best["score"] - runner_score if runner_up else float("inf")

    diagnostics = {
        "status": "accepted",
        "reason": "strong_unique_match",
        "selected_id": best["neuron_id"],
        "selected_label": best["label"],
        "score": best["score"],
        "runner_up_label": runner_up["label"] if runner_up else None,
        "runner_up_score": runner_score,
        "margin": margin,
        "candidates": ranked[:3],
        "thresholds": {
            "minimum_similarity": NEURON_METADATA_MIN_SIMILARITY,
            "minimum_margin": NEURON_METADATA_MIN_MARGIN,
        },
    }

    if best["score"] < NEURON_METADATA_MIN_SIMILARITY:
        diagnostics.update(status="rejected", reason="similarity_below_threshold")
        LAST_METADATA_DECISION = diagnostics
        return None, best["score"]

    if runner_up is not None and margin < NEURON_METADATA_MIN_MARGIN:
        diagnostics.update(status="rejected", reason="ambiguous_top_candidates")
        LAST_METADATA_DECISION = diagnostics
        return None, best["score"]

    LAST_METADATA_DECISION = diagnostics
    return best["neuron_id"], best["score"]


def attention_filter(nn, candidates, threshold=0.5):
    filtered = []
    for nid in candidates:
        neuron = nn.neurons.get(nid)
        if not neuron:
            continue
        score = neuron.soma.confidence + neuron.activation
        if score >= threshold:
            filtered.append(nid)
    return filtered


def region_attention_search(nn, platform, embedding):
    """Use the metadata graph to find eligible identity nodes, not to score images."""
    candidate_ids = set()
    for region_id in sorted(platform.region_refs, key=str):
        region = nn.regions.get(region_id)
        if region is not None:
            candidate_ids.update(region.neurons)
    if not candidate_ids:
        global LAST_METADATA_DECISION
        LAST_METADATA_DECISION = {"status": "rejected", "reason": "no_graph_candidates"}
        return None, 0.0
    return find_best_in_candidates(nn, embedding, candidate_ids)


def update_neuron(nn, nid, embedding, correct: bool):
    neuron = nn.neurons.get(nid)
    if not neuron:
        return
    old = neuron.soma.data.get("embedding")
    if old is not None:
        neuron.soma.data["embedding"] = old * 0.7 + embedding * 0.3
    else:
        neuron.soma.data["embedding"] = embedding
    if correct:
        neuron.soma.confidence = min(1.0, neuron.soma.confidence + 0.05)
        neuron.soma.support += 1
    else:
        neuron.soma.confidence = max(0.0, neuron.soma.confidence - 0.1)
        neuron.soma.mistakes += 1


def predict_face(brain, platform, img_path, true_label=None):
    """Read-only inference: metadata is returned only after a guarded match.

    Passing true_label measures correctness; it does not silently update confidence,
    prototypes, labels, or graph structure.
    """
    emb = get_face_embedding(img_path)
    if emb is None:
        return {"prediction": "NO_FACE", "reason": "no_face_detected"}

    matched_nid, score = region_attention_search(brain, platform, emb)
    decision = dict(LAST_METADATA_DECISION)

    if matched_nid is None:
        return {
            "prediction": "UNKNOWN",
            "reason": decision.get("reason", "metadata_selection_rejected"),
            "similarity": float(score),
            "margin": decision.get("margin"),
            "selection": decision,
        }

    neuron = brain.neurons.get(matched_nid)
    label = neuron.soma.metadata.get("label") if neuron is not None else None
    if not isinstance(label, str) or not label.strip():
        return {
            "prediction": "UNKNOWN",
            "reason": "selected_node_has_no_valid_label",
            "similarity": float(score),
            "selection": decision,
        }

    result = {
        "prediction": label,
        "neuron_id": matched_nid,
        "similarity": float(score),
        "reason": "strong_unique_match",
        "selection": decision,
    }
    if true_label is not None:
        result["correct"] = (label == str(true_label))
    return result


def load_dataset(folder_path):
    X, y = [], []
    for person_name in os.listdir(folder_path):
        person_path = os.path.join(folder_path, person_name)
        if not os.path.isdir(person_path):
            continue
        for img_file in os.listdir(person_path):
            if not img_file.endswith(('.jpg', '.png')):
                continue
            emb = get_face_embedding(os.path.join(person_path, img_file))
            if emb is None:
                continue
            X.append(emb)
            y.append(person_name)
    return np.array(X), np.array(y)


# =========================
# SETUP
# =========================

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
mtcnn = MTCNN(image_size=160, margin=10, device=device)
resnet = InceptionResnetV1(pretrained='vggface2').eval().to(device)


def get_face_embedding(img_path):
    img = Image.open(img_path).convert('RGB')
    face = mtcnn(img)
    if face is None:
        return None
    face = face.unsqueeze(0).to(device)
    with torch.no_grad():
        embedding = resnet(face)
    return embedding.squeeze(0).cpu().numpy()


train_folder = r"C:\Users\mhd_gamer\Documents\GitHub\face_recognition\FACE DATASET\Faces\train"
test_folder  = r"C:\Users\mhd_gamer\Documents\GitHub\face_recognition\FACE DATASET\Faces\test"

brain    = BrainSystem()
platform = NeuralPlatform(name="faces")

X_train, y_train = load_dataset(train_folder)
X_test,  y_test  = load_dataset(test_folder)
print(f"✅ Loaded {len(X_train)} train and {len(X_test)} test images")

# Build metadata nodes by identity and keep every normalized image embedding
# outside the graph. Do not EMA-smear one image into another identity's prototype.
FEATURES_BY_NEURON.clear()
label_to_neuron = {}
for emb, label in zip(X_train, y_train):
    label = str(label)
    if label not in label_to_neuron:
        nid = brain.create_neuron()
        brain.neurons[nid].soma.metadata["label"] = label
        label_to_neuron[label] = nid
        FEATURES_BY_NEURON[nid] = []

    normalized = _normalise_embedding(emb)
    if normalized is not None:
        FEATURES_BY_NEURON[label_to_neuron[label]].append(normalized)

# Keep graph nodes metadata-only.
for neuron in brain.neurons.values():
    neuron.soma.data.pop("embedding", None)

# --- FIX 1 + FIX 2: wire region → brain → platform correctly ---
global_region = Region(name="global")
for nid in brain.neurons:
    global_region.add_neuron(nid)

brain.regions[global_region.uuid]   = global_region   # FIX 1: uuid key, not "global"
brain.platforms[platform.uuid]      = platform         # FIX 2a: register platform
platform.add_region(global_region.uuid)                # FIX 2b: wire region to platform

# Sanity check
print("Neurons:", len(brain.neurons))
print("Regions:", len(brain.regions))
print("Platform refs:", platform.region_refs)

# Test: inference must not mutate metadata or add guessed UNKNOWN identity nodes.
test_img = os.path.join(test_folder, str(y_test[0]), os.listdir(os.path.join(test_folder, str(y_test[0])))[0])
neuron_count_before = len(brain.neurons)
result = predict_face(brain, platform, test_img, true_label=str(y_test[0]))
assert len(brain.neurons) == neuron_count_before, "Inference should not add guessed metadata nodes"
assert all("embedding" not in n.soma.data for n in brain.neurons.values()), "Embeddings must stay outside metadata graph"
print(result)
print("Graph stores labels/metadata only; normalized training features live in FEATURES_BY_NEURON.")


# Metadata-selection diagnostics
print("Metadata selection thresholds:", {
    "minimum_similarity": NEURON_METADATA_MIN_SIMILARITY,
    "minimum_margin": NEURON_METADATA_MIN_MARGIN,
    "top_examples_per_identity": NEURON_METADATA_TOP_EXEMPLARS,
})
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

# ── UI components ─────────────────────────────────────────────────────────────
path_input = widgets.Text(
    placeholder=r"C:\Users\mhd_gamer\Documents\GitHub\face_recognition\FACE DATASET\Faces\test\img1.jpg",
    description="Image path:",
    layout=widgets.Layout(width="600px")
)
run_btn    = widgets.Button(description="Test", button_style="primary")
out        = widgets.Output()

# ── On click ──────────────────────────────────────────────────────────────────
def on_run(b):
    with out:
        clear_output(wait=True)
        img_path = path_input.value.strip()

        if not os.path.isfile(img_path):
            print(f"❌ File not found: {img_path}")
            return

        result = predict_face(brain, platform, img_path)

        label = result["prediction"]
        sim   = result.get("similarity", 0.0)
        if hasattr(sim, "item"):
            sim = sim.item()

        # colour-code by confidence
        if label == "UNKNOWN" or label == "NO_FACE":
            color = "tomato"
        elif sim >= 0.75:
            color = "limegreen"
        else:
            color = "orange"

        # show image + overlay
        img = Image.open(img_path).convert("RGB")
        fig, ax = plt.subplots(figsize=(5, 5))
        ax.imshow(img)
        ax.axis("off")
        ax.set_title(
            f"Prediction : {label}\nSimilarity  : {sim:.3f}",
            fontsize=13, color=color, fontweight="bold",
            pad=10
        )
        patch = mpatches.Patch(color=color, label=label)
        ax.legend(handles=[patch], loc="lower right", fontsize=10)
        plt.tight_layout()
        plt.show()

        print(f"  prediction : {label}")
        print(f"  similarity : {sim:.4f}")
        print(f"  neuron_id  : {result.get('neuron_id', 'N/A')}")

run_btn.on_click(on_run)

# ── Render ────────────────────────────────────────────────────────────────────
display(widgets.HBox([path_input, run_btn]), out)
```
